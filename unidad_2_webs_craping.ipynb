# Web Scraping

Fuentes de datos de texto: sitios web

## Ejercicio 1

```python
# importacion de las librerias que voy a usar
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

### Solicitud de la informacion

```python
sitio = "https://rivaquiroga.github.io/taller-web-scraping-python-2023/ejercicio-1.html"
respuesta = requests.get(sitio)
contenido = respuesta.text
soup = BeautifulSoup(contenido, "html.parser")
```

### Extraemos los datos que nos interesan

```python
# Buscamos los encabezados h2
elementos_h2 = soup.find_all("h2")

# Iteramos sobre la variable elementos_h2
librerias = []
for elemento in elementos_h2:
  librerias.append(elemento.get_text())
```

```python
# Buscamos las descripciones
elementos_p = soup.find_all("p", class_ = "librerias")

# Iteramos sobre la variable elementos_p
descripcion = []
for elemento in elementos_p:
  descripcion.append(elemento.get_text())
```

```python
# Buscamos los enlaces
elementos_a = soup.find_all("a")

# Iteramos sobre la variable elementos_a
enlaces = []
for elemento in elementos_a:
  enlaces.append(elemento.get("href"))
```

### Transformacion de los datos a una estructura data frame

```python
# creamos un diccionario con los lista ya creadas
web_scraping = {
    "libreria": librerias,
    "descripcion": descripcion,
    "enlaces": enlaces
}

# covertimos la variables web_scraping a un data frame
df_librerias = pd.DataFrame(web_scraping).reset_index()

# guardamos el dataframe en un formato CSV
df_librerias.to_csv("librerias_web_scraping.csv", index=False)
```

## Ejercicio 2

### Solicitud de la informacion

```python
sitio = "http://programminghistorian.org/es/lecciones/"
respuesta = requests.get(sitio)
contenido = respuesta.text
soup = BeautifulSoup(contenido, "html.parser")
```

### Extraemos los datos que nos interesan

```python
# cortamos solo la parte de la pagina (info que nos interes)
lista_lecciones = soup.find_all('div', class_ = 'lesson-description')
```

```python
# creamos las variables necesarias que usaremos para recibir los datos
titulos = []
autores = []
resumenes = []
enlaces = []
fechas = []
topicos = []
actividades = []
dificultades = []
```

```python
# iteramos en lista_lecciones, agregando el valor encontrado para cada variable
for leccion in lista_lecciones:
    titulos.append(leccion.h2.get_text(strip = True))
    autores.append(leccion.h3.get_text(strip = True))
    resumenes.append(leccion.p.get_text(strip = True))
    enlace = leccion.a.get("href")
    enlaces.append(f"http://programminghistorian.org{enlace}")
    fechas.append(leccion.find("span", class_ = "date").get_text())
    topicos.append(leccion.find("span", class_ = "topics").get_text().split())
    actividades.append(leccion.find("span", class_ = "activity").get_text())
    dificultades.append(leccion.find("span", class_ = "difficulty").get_text())
```

### Transformacion de los datos a una estructura data frame

```python
# juntamos las listas en un diccionario
lecciones = {
    "titulo": titulos,
    "autores": autores,
    "resumen": resumenes,
    "enlace": enlaces,
    "fecha_publicacion": fechas,
    "topicos": topicos,
    "actividad": actividades,
    "dificultad": dificultades
}

# convertimos el diccionario en un dataset
df_lecciones = pd.DataFrame(lecciones).reset_index()

# convertimos los valores del campo dificultad en integer
df_lecciones['dificultad'] = df_lecciones['dificultad'].astype(int)

# guardamos el dataframe en un formato CSV
df_lecciones.to_csv('lecciones_ph_es.csv', index=False)
```
