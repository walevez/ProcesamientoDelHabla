#Trabajo con documentos que contienen texto.

Cargue el archivo en un repositorio de github para no perder el acceso al mismo en caso de que la pagina cambie el link.

```python
link_doc = 'https://raw.githubusercontent.com/walevez/ProcesamientoDelHabla/refs/heads/main/Nunca-Mas.pdf'
```

### Descarga de bibliotecas necesarias

```python
! pip install PyPDF2
```

```python
# prompt: necesito descrgar el archivo descargado en la variable link_doc
import requests
import os
import PyPDF2

# Toma la ultima parte de a URL, en este caso el nombre del archivo
# filename = 'Nunca-Mas.pdf'
filename = os.path.basename(link_doc)

print(filename)

# Envia la peticion a la url para descargar el archivo
response = requests.get(link_doc)

# Chequea si la request obtuvo como respuesta el codigo de estado 200
if response.status_code == 200:
    # Abre el archivo en modo binario y escrbibe el contenido ahi ('wb')
    with open(filename, 'wb') as f:
        f.write(response.content)
    print(f"File '{filename}' downloaded successfully!")
else:
    # Si el codigo de estado es distinto a 200 (ej. 400, 502 etc)
    # muestra imprime el siguiente mensaje.
    print(f"Failed to download file. Status code: {response.status_code}")
```

```python
# listamos los archivos
! ls -lh
```

```python
# asi como en la funcion anterior el escribimos el archivo en binario,
# ahora lo vamos a leer usando ('rb')
with open(filename, 'rb') as file:
    # Abrimos el archivo file con metodos de la bibliote y los guardamos
    # en readaer.
    reader = PyPDF2.PdfReader(file)

    # inicializamos un varibles tipo string vacia
    full_text = ""

    # iteramos el archivo reader leyendo pagina por pagina, guardando el texo
    # obtenido en la variable full_text recien decladra.
    for page_num in range(len(reader.pages)):
        page = reader.pages[page_num]
        full_text += page.extract_text() + "\n"

# una vez terminada la extraccion de texto, imprimimos el archivo.
print(full_text[:200])
print("\n...")
print(f"Total characters extracted: {len(full_text)}")
```
