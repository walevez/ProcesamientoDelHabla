```python
import pandas as pd
import os

# Define the directory path
directory = 'sentiment labelled sentences'
files = ['amazon_cells_labelled.txt', 'imdb_labelled.txt', 'yelp_labelled.txt']

# Load and concatenate the files
df_list = []
for file in files:
    path = os.path.join(directory, file)
    # These files are tab-separated and have no header
    temp_df = pd.read_csv(path, sep='\t', header=None, names=['sentence', 'label'])
    temp_df['source'] = file.split('_')[0] # Track the source
    df_list.append(temp_df)

# Combine into one dataset
sentiment_df = pd.concat(df_list, ignore_index=True)
```
