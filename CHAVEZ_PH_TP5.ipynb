PREPARACION DEL DATASET

```python
# dataset original: base de conocimiento
preguntas_tp3 = list(dataset_tp3.keys())
respuestas_tp3 = list(dataset_tp3.values())

# dataset de evaluacion
preguntas_tp5 = list(dataset_tp5.keys())
respuestas_tp5 = list(dataset_tp5.values())

print("Registros TP3:", len(preguntas_tp3))
print("Registros TP5:", len(preguntas_tp5))
```

LIBRERIAS NECESARIAS

```python
import chromadb
import torch
from chromadb.utils import embedding_functions
from transformers import AutoTokenizer, AutoModelForCausalLM
```

CARGA DE CHROMADB

```python
# Crear la base de datos vectorial en memoria
cliente = chromadb.EphemeralClient()
```

CARGA DEL MODELO DE EMBEDDING MINILM

```python

# Seleccionar el primer modelo de embeddings
embedding_minilm = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2",
    device="cpu"
)

# Crear la colección
coleccion_minilm = cliente.get_or_create_collection(
    name="molino_minilm",
    embedding_function=embedding_minilm
)

# Guardar las preguntas originales y sus respuestas
coleccion_minilm.upsert(
    ids=[f"doc{i}" for i in range(len(preguntas_tp3))],
    documents=preguntas_tp3,
    metadatas=[
        {"respuesta": respuesta}
        for respuesta in respuestas_tp3
    ]
)

print("Registros MiniLM:", coleccion_minilm.count())
```

CARGA DEL MODELO DE EMBEDDING BGE

```python
embedding_bge = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="BAAI/bge-m3",
    device="cpu"
)

coleccion_bge = cliente.get_or_create_collection(
    name="molino_bge",
    embedding_function=embedding_bge
)

coleccion_bge.upsert(
    ids=[f"doc{i}" for i in range(len(preguntas_tp3))],
    documents=preguntas_tp3,
    metadatas=[
        {"respuesta": respuesta}
        for respuesta in respuestas_tp3
    ]
)

print("Registros BGE-M3:", coleccion_bge.count())
```

VERIFICACION DE GPU EN EL COLAB

```python
# Verificar GPU
dispositivo = "cuda" if torch.cuda.is_available() else "cpu"
tipo_dato = torch.float16 if dispositivo == "cuda" else torch.float32

print("Dispositivo:", dispositivo)
```

CARGA DEL MODELO LLM

```python
modelo_llm = "Qwen/Qwen2.5-1.5B-Instruct"

# Cargar tokenizador
tokenizer = AutoTokenizer.from_pretrained(modelo_llm)

# Cargar Qwen
llm = AutoModelForCausalLM.from_pretrained(
    modelo_llm,
    dtype=tipo_dato,
    device_map=dispositivo
)

print("Qwen cargado correctamente")
```

ARMADO DE CHATBOT

```python
class ChatBot:

    def __init__(self, coleccion, llm, tokenizer):
        self.coleccion = coleccion
        self.llm = llm
        self.tokenizer = tokenizer

    def responder(self, pregunta):

        # Recuperar los 3 contextos más similares
        resultados = self.coleccion.query(
            query_texts=[pregunta],
            n_results=3
        )

        contextos = [
            dato["respuesta"]
            for dato in resultados["metadatas"][0]
        ]

        # Preparar el mensaje para Qwen
        mensajes = [
            {
                "role": "system",
                "content": (
                    "Sos un asistente de atención al cliente de un molino harinero. "
                    "Respondé en español de forma breve y precisa. "
                    "Utilizá exclusivamente la información del contexto proporcionado. "
                    "Si preguntan precios, informá los precios individuales con sus impuestos. "
                    "No calcules importes totales si no conocés las cantidades. "
                    "No inventes descuentos, promociones, requisitos ni condiciones comerciales. "
                    "No agregues información que no sea necesaria para responder. "
                    "Si el contexto no contiene la respuesta, indicá que no disponés de esa información."
                )
            },
            {
                "role": "user",
                "content": (
                    f"Contexto:\n{chr(10).join(contextos)}\n\n"
                    f"Pregunta: {pregunta}"
                )
            }
        ]

        entradas = self.tokenizer.apply_chat_template(
            mensajes,
            add_generation_prompt=True,
            return_tensors="pt",
            return_dict=True
        ).to(self.llm.device)

        with torch.inference_mode():
            salida = self.llm.generate(
                **entradas,
                max_new_tokens=100,
                do_sample=False,
                pad_token_id=self.tokenizer.eos_token_id
            )

        tokens = salida[0][entradas["input_ids"].shape[-1]:]

        respuesta = self.tokenizer.decode(
            tokens,
            skip_special_tokens=True
        )

        return respuesta, contextos
```

CHATBOTS CON LOS EMBEDDING

```python
chatbot_minilm = ChatBot(coleccion_minilm, llm, tokenizer)
chatbot_bge = ChatBot(coleccion_bge, llm, tokenizer)

print("Ambos chatbots creados correctamente")
```

PRUEBA DE LOS DOS CHATBOTS

```python
pregunta = "¿Cuánto cuesta comprar harina 000 y 0000?"

# Prueba con MiniLM
respuesta_minilm, contextos_minilm = chatbot_minilm.responder(pregunta)

print("=== MiniLM + Qwen ===")
print("Respuesta:", respuesta_minilm)
print("Contextos:", contextos_minilm)

# Prueba con BGE-M3
respuesta_bge, contextos_bge = chatbot_bge.responder(pregunta)

print("\n=== BGE-M3 + Qwen ===")
print("Respuesta:", respuesta_bge)
print("Contextos:", contextos_bge)
```

d) Evaluacion y comparacion de los modelos de embeddings

En esta etapa se evaluan los dos modelos de embeddings utilizando las 21 preguntas del dataset de prueba. Se analizan los resultados de recuperacion y las respuestas generadas por Qwen.

```python
import pandas as pd

def evaluar_recuperacion(coleccion, preguntas):
    resultados = []

    for i, pregunta in enumerate(preguntas):
        id_correcto = f"doc{i}"

        busqueda = coleccion.query(
            query_texts=[pregunta],
            n_results=3
        )

        ids = busqueda["ids"][0]

        resultados.append({
            "Pregunta": pregunta,
            "Hit@1": int(ids[0] == id_correcto),
            "Hit@3": int(id_correcto in ids)
        })

    return pd.DataFrame(resultados)


evaluacion_minilm = evaluar_recuperacion(
    coleccion_minilm, preguntas_tp5
)

evaluacion_bge = evaluar_recuperacion(
    coleccion_bge, preguntas_tp5
)

resumen = pd.DataFrame({
    "Modelo": ["MiniLM", "BGE-M3"],
    "Hit@1 (%)": [
        evaluacion_minilm["Hit@1"].mean() * 100,
        evaluacion_bge["Hit@1"].mean() * 100
    ],
    "Hit@3 (%)": [
        evaluacion_minilm["Hit@3"].mean() * 100,
        evaluacion_bge["Hit@3"].mean() * 100
    ]
})

display(resumen.round(2))
```

```python
# @title
def probar_chatbot(chatbot, dataset):
    resultados = []

    for numero, (pregunta, esperada) in enumerate(dataset.items(), 1):
        generada, contextos = chatbot.responder(pregunta)

        resultados.append({
            "pregunta": pregunta,
            "esperada": esperada,
            "generada": generada,
            "contextos": contextos
        })

        print(f"Pregunta {numero}/{len(dataset)} completada")

    return resultados


resultados_minilm = probar_chatbot(chatbot_minilm, dataset_tp5)
resultados_bge = probar_chatbot(chatbot_bge, dataset_tp5)
```

```python
comparacion = pd.DataFrame({
    "Pregunta": [r["pregunta"] for r in resultados_minilm],
    "Respuesta esperada": [r["esperada"] for r in resultados_minilm],
    "MiniLM + Qwen": [r["generada"] for r in resultados_minilm],
    "BGE-M3 + Qwen": [r["generada"] for r in resultados_bge]
})

display(comparacion)
```

Evaluacion de las respuestas generadas

Se calcula la similitud semántica entre las respuestas generadas por Qwen y las respuestas esperadas del dataset de evaluacion, para comparar el rendimiento de MiniLM y BGE-M3.

```python
from sentence_transformers import SentenceTransformer
from sentence_transformers import util

modelo_evaluador = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

def evaluar_respuestas(resultados):
    evaluacion = []

    for r in resultados:
        vectores = modelo_evaluador.encode(
            [r["esperada"], r["generada"]],
            convert_to_tensor=True
        )

        similitud = util.cos_sim(
            vectores[0], vectores[1]
        ).item()

        evaluacion.append({
            "Pregunta": r["pregunta"],
            "Similitud": similitud
        })

    return pd.DataFrame(evaluacion)


evaluacion_respuestas_minilm = evaluar_respuestas(resultados_minilm)
evaluacion_respuestas_bge = evaluar_respuestas(resultados_bge)

print("MiniLM:", round(evaluacion_respuestas_minilm["Similitud"].mean(), 3))
print("BGE-M3:", round(evaluacion_respuestas_bge["Similitud"].mean(), 3))
```

Conclusión del punto d)

Se evaluaron las respuestas generadas por Qwen utilizando los dos modelos de embeddings y las 21 preguntas del dataset de evaluacion.

MiniLM obtuvo una similitud semantica promedio de 0,652, mientras que BGE-M3 alcanzó 0,725.

Tambien se observó que BGE-M3 recupero con mayor frecuencia el documento correcto en primera posicion.

Por estos resultados, elegiria BGE-M3 para el chatbot del molino harinero, aunque durante las pruebas se detectaron respuestas generadas con informacion incorrecta.

e) Evaluacion de Context Precision y Context Recall

Se implementan manualmente metricas de recuperacion basadas en la identificacion y posicion de los documentos relevantes, sin utilizar la libreria RAGAS.

```python
def evaluar_contextos(coleccion, preguntas):
    resultados = []

    for i, pregunta in enumerate(preguntas):

        # Documento correcto según el dataset TP3
        id_correcto = f"doc{i}"

        # Recuperar los tres documentos más similares
        consulta = coleccion.query(
            query_texts=[pregunta],
            n_results=3
        )

        ids = consulta["ids"][0]

        # Relevancia de cada contexto
        relevancia = [
            int(doc_id == id_correcto)
            for doc_id in ids
        ]

        # Context Precision: Average Precision@3
        aciertos = 0
        suma_precision = 0

        for posicion, relevante in enumerate(relevancia, start=1):
            if relevante:
                aciertos += 1
                suma_precision += aciertos / posicion

        context_precision = suma_precision / 1

        # Context Recall: único documento relevante
        context_recall = int(id_correcto in ids)

        resultados.append({
            "Pregunta": pregunta,
            "Context Precision": context_precision,
            "Context Recall": context_recall
        })

    return pd.DataFrame(resultados)
```

```python
metricas_minilm = evaluar_contextos(
    coleccion_minilm, preguntas_tp5
)

metricas_bge = evaluar_contextos(
    coleccion_bge, preguntas_tp5
)

comparacion_metricas = pd.DataFrame({
    "Modelo": ["MiniLM", "BGE-M3"],
    "Context Precision (%)": [
        metricas_minilm["Context Precision"].mean() * 100,
        metricas_bge["Context Precision"].mean() * 100
    ],
    "Context Recall (%)": [
        metricas_minilm["Context Recall"].mean() * 100,
        metricas_bge["Context Recall"].mean() * 100
    ]
})

display(comparacion_metricas.round(2))
```

Conclusión del punto e)

Para evaluar la recuperacion de informacion, se implementaron manualmente métricas basadas en la posicion del documento relevante y su presencia entre los tres primeros resultados.

BGE-M3 obtuvo un valor de 88,10 % en la métrica de precision, frente al 81,75 % de MiniLM. En cambio, MiniLM alcanzo un recall de 95,24 %, superando el 90,48 % de BGE-M3.

Estos resultados refuerzan parcialmente la elección de BGE-M3, porque posiciona mejor los documentos relevantes, aunque MiniLM recupera mas documentos esperados entre los tres primeros resultados.

Las metricas utilizadas son aproximaciones basadas en identificadores de documentos y no equivalen a las evaluaciones semanticas de RAGAS.
